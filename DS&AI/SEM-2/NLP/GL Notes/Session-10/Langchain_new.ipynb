### 🔍 Pipeline Flow

Query

↓

Similarity Search (with score)

↓

Filter (threshold)

↓

Fallback (best doc)

↓

Context

↓

GPT-2 Answer

```python
from langchain_community.document_loaders import WebBaseLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

from transformers import pipeline
from langchain_huggingface import HuggingFacePipeline
```

```python
loader = WebBaseLoader("https://www.mygreatlearning.com/blog/machine-learning-models/")
data = loader.load()
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

all_splits = text_splitter.split_documents(data)
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-mpnet-base-v2"
)
```

```python
vectorstore = FAISS.from_documents(
    documents=all_splits,
    embedding=embeddings
)
```

```python
query = "What is Machine Learning?"
docs = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(docs):
    print(f"\n--- Document {i+1} ---\n")
    print(doc.page_content[:500])
```

```python
query = "What is Deep Learning?"
docs = vectorstore.similarity_search(query, k=3)

for i, doc in enumerate(docs):
    print(f"\n--- Document {i+1} ---\n")
    print(doc.page_content[:500])
```

```python
query = "What is RAG model"


docs_with_scores = vectorstore.similarity_search_with_score(query, k=3)

THRESHOLD = 0.5   # you can tune this

relevant_docs = []
for doc, score in docs_with_scores:
    if score < THRESHOLD:   # lower score = better match (FAISS distance)
        relevant_docs.append(doc)

if len(relevant_docs) == 0:
    print("I don’t have enough information to answer this question.")
else:
    context = "\n\n".join([doc.page_content for doc in relevant_docs])
    print(f"\n\nContext:\n{context}")
```

```python
query = "What is RAG model"
docs_with_scores = vectorstore.similarity_search_with_score(query, k=3)


THRESHOLD = 1.5   # tune this based on your scores

relevant_docs = []
for doc, score in docs_with_scores:
    print(f"Score: {score}")  # debug
    if score < THRESHOLD:
        relevant_docs.append(doc)


if len(relevant_docs) == 0:
    print("⚠️ No strong match found. Using best available context.\n")
    relevant_docs = [docs_with_scores[0][0]]   # take best match


context = "\n\n".join([doc.page_content for doc in relevant_docs])
```

```python
from transformers import pipeline
from langchain_huggingface import HuggingFacePipeline

pipe = pipeline(
    "text-generation",
    model="gpt2",
    max_new_tokens=120,
    temperature=0.7,
    do_sample=True
)

llm = HuggingFacePipeline(pipeline=pipe)
```

```python
prompt = f"""
Use ONLY the information from the context below.

If the answer is not present, say:
"I don’t have enough information."

Context:
{context}

Question: {query}

Answer in a clear and complete sentence:
"""
```

```python
response = llm.invoke(prompt)

# Clean output
answer = response.replace(prompt, "").strip()

print("\nAnswer:\n", answer)
```
